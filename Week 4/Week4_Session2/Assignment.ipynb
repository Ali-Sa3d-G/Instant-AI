Q6.
Write a program that takes a month number (1–12) and prints(use if and match):

"Winter"

"Spring"

"Summer"

"Autumn"

````python
month = int(input("Enter a month: "))
match month:
  case 12 | 1 | 2:
    print("Winter")
  case 3 | 4 | 5:
    print("Spring")
  case 6 | 7 | 8:
    print("Summer")
  case 9 | 10 | 11:
    print("Autumn")
  case _:
    print("Invalid month entered.")
````

````python
month = int(input("Enter a month: "))
if month in [12,1,2]:
  print("Winter")
elif month in [3,4,5]:
  print("Spring")
elif month in [6,7,8]:
    print("Summer")
elif month in [9,10,11]:
  print("Autumn")
else:
  print("Invalid month entered")
````

````python
month = int(input("Enter a month: "))
seasons =[
    ([12, 1, 2], "Winter"),
    ([3, 4, 5], "Spring"),
    ([6, 7, 8], "Summer"),
    ([9, 10, 11], "Autumn")]

found = False
for month_range, season_name in seasons:
    if month in month_range:
        print(season_name)
        found = True
        break

if not found:
    print("Invalid month entered.")
````

Write a program that:

Takes an integer number from the user

Calculates its factorial using a for loop

Prints the result

Example:

Input → 5

Output → 120

````python
num = int(input("Enter a number"))
factorial = 1
for i in range(num):
  factorial *= i+1
print(f"The factorial of {num} is {factorial}")
````

````python
num = int(input("Enter a number"))
factorial = 1
i = 1
while i <= num :
  factorial *= i
  i+=1
print(f"The factorial of {num} is {factorial}")
````

Q9*
Given a list of numbers:

if number is even → add it to even_list

otherwise → add it to odd_list
Print both lists at the end.

````python
lst = [1, 2, 3, 4, 5, 6, 7, 8]
even_lst = []
odd_lst = []

for i in range(len(lst)):
  if lst[i] % 2 == 0:
    even_lst.append(lst[i])
  else:
    odd_lst.append(lst[i])
print(f"Even numbers: {even_lst}")
print(f"Odd numbers: {odd_lst}")
````

````python
lst = [1, 2, 3, 4, 5, 6, 7, 8]

even_lst = [num for num in lst if num % 2 == 0]
odd_lst = [num for num in lst if num % 2 != 0]

print(f"Even numbers (using list comprehension): {even_lst}")
print(f"Odd numbers (using list comprehension): {odd_lst}")
````

Q10.
Take a list of words.
Count how many words have length > 5.

````python
lst = ["Programming", "Python", "Ai", "Data", "Analysis"]
words = []
count = 0

for i in range(len(lst)):
  if len(lst[i]) > 5:
    words.append(lst[i])
    count += 1
print(f"Number of words = {count}")
print(words)
````

````python
lst = ["Programming", "Python", "Ai", "Data", "Analysis"]
words = []
count = 0

words = [word for word in lst if len(word) > 5]
count = len(words)
print(f"Number of words = {count}")
print(words)
````

Level 4 – Advanced Thinking with for Loop

Write a program that:

Checks if a number is prime using a for loop

````python
num = int(input("Enter a number: "))
for i in range(2, num):
  if num % i == 0:
    print(f"{num} is not a prime number")
    break
else:
  print(f"{num} is a prime number")
````

````python
def is_prime(num):
  if num <= 0:
    return False

  for i in range(2, num):
    if num % i == 0:
      return False
  return True

num = int(input("Enter a number: "))

if is_prime(num):
  print(f"{num} is a prime number")
else:
  print(f"{num} is not a prime number")
````

Given a number N. Print a pyramid that has N rows.

For more clarification see the example below.

Input

Only one line containing a number N (1 ≤ N ≤ 99).

Output

Print the answer according to the required above.

Note
Don't print any extra spaces after symbol " * ".

Input

4

Output


```
   *
  ***
 *****
*******



```


````python
rows = int(input("Enter the number of rows: "))
for i in range(rows):
    for j in range(rows - i - 1):
        print(" ", end="")
    for k in range(2 * i + 1):
        print("*", end="")
    print()
````

Given a number N. Print a diamond that has 2N rows.

For more clarification see the example below.

Input
Only one line containing number N (1 ≤ N ≤ 99).

Output
Print the answer according to the required above.

Example
Input

4

Output
```
   *
  ***
 *****
*******
*******
 *****
  ***
   *
```
Note

Don't print any extra spaces after symbol " * ".

````python
rows = int(input("Enter the number of rows: "))

# Upper part of the diamond
for i in range(rows):
    for j in range(rows - i - 1):
        print(" ", end="")
    for k in range(2 * i + 1):
        print("*", end="")
    print()

# Lower part of the diamond
for i in range(rows - 2, -1, -1):
    for j in range(rows - i - 1):
        print(" ", end="")
    for k in range(2 * i + 1):
        print("*", end="")
    print()
````
