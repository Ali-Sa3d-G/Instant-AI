# Q 6

```python
month = int(input("Enter a month: "))
match month:
  case 12 | 1 | 2:
    print("Winter")
  case 3 | 4 | 5:
    print("Spring")
  case 6 | 7 | 8:
    print("Summer")
  case 9 | 10 | 11:
    print("Autumn")
  case _:
    print("Invalid month entered.")
```

```python
month = int(input("Enter a month: "))
if month in [12,1,2]:
  print("Winter")
elif month in [3,4,5]:
  print("Spring")
elif month in [6,7,8]:
    print("Summer")
elif month in [9,10,11]:
  print("Autumn")
else:
  print("Invalid month entered")
```

```python
month = int(input("Enter a month: "))
seasons =[
    ([12, 1, 2], "Winter"),
    ([3, 4, 5], "Spring"),
    ([6, 7, 8], "Summer"),
    ([9, 10, 11], "Autumn")]

found = False
for month_range, season_name in seasons:
    if month in month_range:
        print(season_name)
        found = True
        break

if not found:
    print("Invalid month entered.")
```

# Q 9

```python
lst = [1, 2, 3, 4, 5, 6, 7, 8]
even_lst = []
odd_lst = []

for i in range(len(lst)):
  if lst[i] % 2 == 0:
    even_lst.append(lst[i])
  else:
    odd_lst.append(lst[i])
print(f"Even numbers: {even_lst}")
print(f"Odd numbers: {odd_lst}")
```

```python
lst = [1, 2, 3, 4, 5, 6, 7, 8]

even_lst = [num for num in lst if num % 2 == 0]
odd_lst = [num for num in lst if num % 2 != 0]

print(f"Even numbers (using list comprehension): {even_lst}")
print(f"Odd numbers (using list comprehension): {odd_lst}")
```

```python
num = int(input("Enter a number: "))
for i in range(2, num):
  if num % i == 0:
    print(f"{num} is not a prime number")
    break
else:
  print(f"{num} is a prime number")
```

```python
def is_prime(num):
  if num <= 0:
    return False

  for i in range(2, num):
    if num % i == 0:
      return False
  return True

num = int(input("Enter a number: "))

if is_prime(num):
  print(f"{num} is a prime number")
else:
  print(f"{num} is not a prime number")
```

```python
rows = int(input("Enter the number of rows: "))
for i in range(rows):
    for j in range(rows - i - 1):
        print(" ", end="")
    for k in range(2 * i + 1):
        print("*", end="")
    print()
```

# Q5

```python
rows = int(input("Enter the number of rows: "))

# Upper part of the diamond
for i in range(rows):
    for j in range(rows - i - 1):
        print(" ", end="")
    for k in range(2 * i + 1):
        print("*", end="")
    print()

# Lower part of the diamond
for i in range(rows - 2, -1, -1):
    for j in range(rows - i - 1):
        print(" ", end="")
    for k in range(2 * i + 1):
        print("*", end="")
    print()
```
